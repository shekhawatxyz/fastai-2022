```
input frame + action -> pretrained predictor -> baseline prediction
         |                                      |
         +--> raw effect statistics ------------+--> scope ensemble -> prediction
                       |                              ^
                       +--> infer behavioral roles ---+
                       ^
                       +------ observed outcome (after prediction)
```

The observations remain unchanged. Re-index the sufficient statistics through newly inferred roles. Scope experts retain different interpretations rather than force all prediction through one ontology. This is a bounded residual, not a full self-reorganizing world model.

Textual motivation: [Vākyapadīya 1.84](https://www.wisdomlib.org/hinduism/book/vakyapadiya-of-bhartrihari/d/doc1216529.html), [1.85](https://www.wisdomlib.org/hinduism/book/vakyapadiya-of-bhartrihari/d/doc1216530.html), [2.13](https://www.wisdomlib.org/hinduism/book/vakyapadiya-of-bhartrihari/d/doc1239129.html), and the competing Prābhākara/Bhāṭṭa/Nyāya accounts of compositional meaning ([SEP](https://plato.stanford.edu/entries/language-india/)).
